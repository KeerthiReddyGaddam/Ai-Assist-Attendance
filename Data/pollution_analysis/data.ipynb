```python
df = pd.read_csv("C:\Users\keert\OneDrive\Desktop\DAV\Data\pollution_analysis\global air pollution dataset.csv")
aqi_category = df["AQI Category"].value_count().reset_index()
aqi_category.colums = ["AQI Category","Count"]
print(aqi_category.head())
fig = px.pie(
    aqi_category,
    names = "AQI Category",
    values = "Count",
    title = "Distribution of AQI Categories",
    hole = 0.3
)

fig.show()
```

```python
fig = px.histogram(
    df,
    x = "AQI Value",
    title = "AQI Distribution",
    nbins = 10
)

fig.show()
```

```python
# scatter plot
fig = px.scatter(
    df,
    x = "PM2.5 AQI Value",
    y = "AQI Value",
    color = "AQI Category",
    hover_name = "City",
    hover_data = ["Country"],
    title = "PM2.5 AQI vs Overall AQI"
)

fig.show()
```

```python
# heatmap
corr = df[
    ["AQI Value",
    "CO AQI Value",
    "Ozone AQI Value",
    "NO2 AQI Value",
    "PM2.5 AQI Value"
]].corr()

fig = px.imshow(
    corr,
    text_suto = True,
    title = "AQI and Pollutant Correlation"
)
fig.show()
```
